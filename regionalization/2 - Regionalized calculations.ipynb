# Inclusion of spatial data during LCA calculations (regionalization)

The team at Cauldron have been at the forefront of research on method development and implementation of regionalization in LCA for 15 years. During that time, we have investigated a number of different approaches, and have consolidated on a practical, performant, and parsimonious approach for including spatial data.

We have a fundamental problem in LCA - the spatial scale of our inventory doesn't line up with the spatial scale of our impact assessment. For example, here are the states in Malaysia, and the WWF ecoregions:

![image](img/compare.png)

We could just look at how much land is in the intersection of each state with each ecoregion, but that is usually very wrong - most countries have activities concentrated in certain areas, and a lot of undisturbed land (think e.g. of Canada).

Instead, we can take remote sensing or similar data on where activity is occurring, and use that data as weights to created weighted averages of the characterization factors in each inventory spatial unit. 

Selected cites on regionalization in LCA:

* [Matrix-based Methods for Regionalized Life Cycle Assessment](https://eartharxiv.org/repository/view/5270/). 2023. Mutel, C., Hellweg, S. EarthArxiv.
* [GIS-based regionalized life cycle assessment: how big is small enough? Methodology and case study of electricity generation](https://pubs.acs.org/doi/abs/10.1021/es203117z). 2011. Mutel, C., Pfister, S., Hellweg, S. Env. Sci. & Tech.

## Case Study: Land Use for palm plantations in Malaysia

Here is [an example of remote sensing data being used in a validated model](https://essd.copernicus.org/articles/16/5111/2024/) to map palm oil:

![image](img/remote.png)

In this case, the purple is high-intensity commercial plantations, and the green is smaller farms.

I have already loaded and processed the GIS data in a separate notebook. Data sources:

* [GLAM land use method](https://zenodo.org/records/10114493)
* [Palm oil maps](https://essd.copernicus.org/articles/16/5111/2024/)
* [WWF ecoregions map](https://www.worldwildlife.org/publications/terrestrial-ecoregions-of-the-world)

Here we will focus on the regionalized calculations:

```python
import bw2data as bd
import bw2regional as bwr
import geopandas as gp
from pathlib import Path
```

```python
PROJECT_NAME = "Palm-oil"
```

```python
bd.projects.set_current(PROJECT_NAME)
```

```python
bd.Database("Malaysian palm land use").set_geocollections()
```

```python
product = bd.get_node(
    name="Palm fruit production in Malaysia - 2022 market mix",
    database="Malaysian palm land use"
)
```

```python
lca = bwr.ExtensionTablesLCA(
    demand={product: 1},
    method=("GLAM", "land use"),
    xtable='regions - ecoregions - palm',
)
lca.lci()
lca.lcia()
```

```python
lca.geodataframe_xtable_spatial_scale().explore(column='score_rel', tiles="OpenStreetMap")
```
